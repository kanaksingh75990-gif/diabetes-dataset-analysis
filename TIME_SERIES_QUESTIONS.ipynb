```python


# Q1.  Create a Python datetime object representing 15 February 2026 at 10:30 AM. Display the year, month, day, hour and minute separately.

from datetime import datetime


dt = datetime(2026, 2, 15, 10, 30)

print("Datetime:", dt)
print("Year:", dt.year)
print("Month:", dt.month)
print("Day:", dt.day)
print("Hour:", dt.hour)
print("Minute:", dt.minute)
```

```python
#Q2.  Create two datetime objects representing 10:00 AM and 12:30 PM on 15 February 2026. Calculate the difference between them in hours and minutes.
from datetime import datetime

# Create datetime objects (12:30 PM in 24-hour format is hour=12)
dt1 = datetime(2026, 2, 15, 10, 0)
dt2 = datetime(2026, 2, 15, 12, 30)

# Calculate the timedelta difference
diff = dt2 - dt1

# Extract hours and minutes
total_seconds = int(diff.total_seconds())
hours = total_seconds // 3600
minutes = (total_seconds % 3600) // 60

print(f"Difference: {hours} hours and {minutes} minutes")
print(f"Total minutes: {total_seconds // 60}")
```

```python
#Q5 Handle Invalid and Missing Dates with errors="coerce"
import pandas as pd

dates = ["2026-02-01", "invalid", "2026-02-05", None]
s = pd.to_datetime(pd.Series(dates), errors="coerce")

missing_mask = s.isna()
missing_indices = s[missing_mask].index.tolist()

print("Converted Series:")
print(s)
print("\nMissing values mask:")
print(missing_mask)
print(f"\nMissing count: {missing_mask.sum()} at indices {missing_indices}")
```

```python
#Q6
import pandas as pd

dates = pd.to_datetime([
    "2026-02-01", "2026-02-02",
    "2026-02-04", "2026-02-06",
    "2026-02-08"
])
sales = pd.Series([100, 150, 120, 180, 200], index=dates, name="units_sold")
print(sales.index)
print(type(sales.index))
```

```python
#Q7. Retrieve Sales Value on 4 February 2026 Using .loc[]
val_feb4 = sales.loc["2026-02-04"]
print("Sales on 2026-02-04:", val_feb4)
```

```python
#Q8. Retrieve First and Last Observations Using .iloc[]
first_obs = sales.iloc[0]
last_obs = sales.iloc[-1]

print(f"First observation: {first_obs} ({sales.index[0].strftime('%Y-%m-%d')})")
print(f"Last observation: {last_obs} ({sales.index[-1].strftime('%Y-%m-%d')})")
```

```python
#9 Select Sales Between 2 February and 6 February 2026 (Inclusive) With .loc[] on a sorted DatetimeIndex, date slices include both start and end labels:
slice_feb2_feb6 = sales.loc["2026-02-02":"2026-02-06"]
print(slice_feb2_feb6)
```

```python
#10 Total and Average Units Sold During February 2026
feb_2026 = sales.loc["2026-02"]
total_units = feb_2026.sum()
avg_units = feb_2026.mean()

print("Total units sold:", total_units)
print("Average units sold:", avg_units)
```

```python
#Q16. Five Business Days Starting from 6 February 2026 (freq="B")

import pandas as pd

b_days = pd.date_range(start="2026-02-06", periods=5, freq="B")
all_span_days = pd.date_range(start=b_days[0], end=b_days[-1], freq="D")
excluded_days = all_span_days.difference(b_days)

print("Business days:")
for d in b_days:
    print(f"  {d.strftime('%Y-%m-%d')} ({d.day_name()})")

print("\nExcluded dates:")
for d in excluded_days:
    print(f"  {d.strftime('%Y-%m-%d')} ({d.day_name()})")
```

```python
import pandas as pd

dates = pd.date_range("2026-02-01", periods=6, freq="D")
daily_sales = pd.Series([100, 120, 150, 130, 180, 200], index=dates)
```

```python
#Q17. Previous Day's Sales Alongside Current Day's Sales (shift(1))
df_q17 = pd.DataFrame({
    "current_sales": daily_sales,
    "previous_day_sales": daily_sales.shift(1)
})
print(df_q17)
```

```python
#Q18. Next Day's Sales Alongside Current Day's Sales (shift(-1)) shift(-1) pulls data up by 1 period so each row's next_day_sales holds the following day's value (NaN on the last day).
df_q18 = pd.DataFrame({
    "current_sales": daily_sales,
    "next_day_sales": daily_sales.shift(-1)
})
print(df_q18)
```

```python
#Q19. Day-to-Day Change in Units Sold Subtracting daily_sales.shift(1) from daily_sales computes the daily change (equivalent to daily_sales.diff()).
day_to_day_change = daily_sales - daily_sales.shift(1)

df_q19 = pd.DataFrame({
    "current_sales": daily_sales,
    "previous_day_sales": daily_sales.shift(1),
    "day_to_day_change": day_to_day_change
})
print(df_q19)
```

```python
#Q20 Q20. Shift the Date Index Forward by 2 Calendar Days (freq="D") Passing freq="D" to .shift() moves the index timestamps forward by 2 days while keeping every sales value attached to its shifted date (no NaN rows are introduced and no observations are dropped).
shifted_index_series = daily_sales.shift(2, freq="D")

print("Original Series:")
print(daily_sales)

print("\nShifted Index Series (shift(2, freq='D')):")
print(shifted_index_series)
```
